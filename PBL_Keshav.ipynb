```python
import re
import hashlib
from cryptography.fernet import Fernet
from datetime import datetime

# -----------------------------
# AI-LIKE HUMAN ERROR DETECTION
# -----------------------------
def detect_human_errors(message, recipient, password):
    warnings = []

    # Strong password check
    if len(password) < 8 or \
       not re.search(r"[A-Z]", password) or \
       not re.search(r"[a-z]", password) or \
       not re.search(r"[0-9]", password) or \
       not re.search(r"[!@#$%^&*]", password):
        warnings.append("⚠ Weak password detected. Use stronger password.")

    # Sensitive keyword detection
    sensitive_words = ["password", "bank", "otp", "pin", "account"]
    if any(word in message.lower() for word in sensitive_words):
        warnings.append("⚠ Sensitive information detected in message.")

    # Empty recipient
    if not recipient.strip():
        warnings.append("⚠ Invalid recipient.")

    return warnings


# -----------------------------
# ENCRYPTION MODULE (AES - Fernet)
# -----------------------------
def generate_key_from_password(password):
    # Derive key from password using SHA-256
    hashed = hashlib.sha256(password.encode()).digest()
    return Fernet(base64.urlsafe_b64encode(hashed))


def encrypt_message(message, fernet_obj):
    return fernet_obj.encrypt(message.encode())


def decrypt_message(cipher_text, fernet_obj):
    return fernet_obj.decrypt(cipher_text).decode()


# -----------------------------
# ANOMALY DETECTION MODULE
# -----------------------------
def detect_anomaly(message_history, new_message):
    if message_history.count(new_message) >= 3:
        return "⚠ Possible replay/spam behavior detected."

    if len(new_message) > 500:
        return "⚠ Abnormally large message size detected."

    return None


# -----------------------------
# LOGGING SYSTEM (AI Monitoring)
# -----------------------------
def log_event(event):
    with open("security_log.txt", "a") as log:
        log.write(f"{datetime.now()} - {event}\n")


# -----------------------------
# MAIN SYSTEM
# -----------------------------
def main():
    print("=== AI-Assisted Secure Messaging System ===\n")

    recipient = input("Enter recipient name: ")
    message = input("Enter message: ")
    password = input("Create login password: ")

    # Step 1: Human Error Detection
    warnings = detect_human_errors(message, recipient, password)

    if warnings:
        print("\n--- WARNINGS DETECTED ---")
        for w in warnings:
            print(w)
            log_event(w)
        print("-------------------------\n")

    # Step 2: Generate key from password
    import base64
    hashed = hashlib.sha256(password.encode()).digest()
    key = base64.urlsafe_b64encode(hashed)
    f = Fernet(key)

    # Step 3: Encrypt message
    encrypted_msg = encrypt_message(message, f)
    print("\nEncrypted Message Sent Securely ✔")
    log_event("Message encrypted and sent.")

    # Simulated history
    history = ["hello", "test", message]

    # Step 4: Anomaly detection
    anomaly = detect_anomaly(history, message)
    if anomaly:
        print(anomaly)
        log_event(anomaly)

    # Step 5: Decrypt message
    decrypted_msg = decrypt_message(encrypted_msg, f)

    print("\n--- Receiver Side ---")
    print("Message Decrypted Successfully ✔")
    print("Received Message:", decrypted_msg)
    log_event("Message decrypted successfully.")


if __name__ == "__main__":
    main()
```
