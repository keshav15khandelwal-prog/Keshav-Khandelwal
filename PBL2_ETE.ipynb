```python
from Cryptodome.Cipher import AES
from Cryptodome.Random import get_random_bytes
import base64


# -------------------------------
# Padding Function (PKCS7)
# -------------------------------
def pad(data):
    pad_len = 16 - len(data) % 16
    return data + bytes([pad_len] * pad_len)


# -------------------------------
# Generate AES-256 Key
# -------------------------------
def generate_key():
    return get_random_bytes(32)   # 256-bit key


# -------------------------------
# Encrypt Message
# -------------------------------
def encrypt_message(message, key):
    iv = get_random_bytes(16)   # 16-byte IV

    cipher = AES.new(key, AES.MODE_CBC, iv)

    padded_message = pad(message.encode())

    ciphertext = cipher.encrypt(padded_message)

    # Combine IV + ciphertext
    encrypted = base64.b64encode(iv + ciphertext).decode()

    return encrypted


# -------------------------------
# MAIN (Sender)
# -------------------------------
if __name__ == "__main__":
    message = input("Enter message: ")

    key = generate_key()

    print("\nAES-256 Key:", key)

    encrypted_msg = encrypt_message(message, key)

    print("\nEncrypted Message:", encrypted_msg)
```
